```python
import pandas as pd

df = pd.read_csv("/content/drive/My Drive/PPWA/report/Tugas-PPWA/Hasil_Prepros.csv")
df.head()
```

Label Encoder

```python
from sklearn.preprocessing import LabelEncoder

# Transformasi data kategorik
label_encoder = LabelEncoder()
df.loc[:, 'kategori_encoded'] = label_encoder.fit_transform(df['kategori'])

# Menampilkan nilai sebelum dan sesudah konversi
print("\nNilai sebelum dan sesudah konversi:")
print(dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))

print("\nDataFrame setelah Label Encoding:")
df
```

Spliting Data

---



```python
from sklearn.model_selection import train_test_split

# Split data
x = df ['stopword_removal']
y = df ['kategori_encoded']

print(x)
print(y)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=12)

print("="*50)
print("Jumlah data latih:", len(x_train))
print("Jumlah data uji:", len(x_test))
```

**TF-IDF (Term Frequency-Inverse Document Frequency)**


---


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Menginisialisasi TfidfVectorizer
vectorizer = TfidfVectorizer()

# Menghitung TF-IDF
tfidf_matrix = vectorizer.fit_transform(x_train)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Inisialisasi TF-IDF Vectorizer
tfidf = TfidfVectorizer()

# Fit dan transform pada data training
x_train_tfidf = tfidf.fit_transform(x_train)

# Transform pada data testing
x_test_tfidf = tfidf.transform(x_test)

# Mendapatkan nama fitur dari TF-IDF
feature_names = tfidf.get_feature_names_out()

# Konversi TF-IDF hasil training ke DataFrame
df_train_tfidf = pd.DataFrame(x_train_tfidf.toarray(), columns=feature_names)

# Konversi TF-IDF hasil testing ke DataFrame
df_test_tfidf = pd.DataFrame(x_test_tfidf.toarray(), columns=feature_names)

# Tampilkan hasil TF-IDF dalam DataFrame
print("TF-IDF untuk data training:")
df_train_tfidf
```

Reduksi dimensi

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
```

```python
# Menggunakan SVD untuk reduksi dimensi
n_components = 80
svd = TruncatedSVD(n_components=n_components)
svd_matrix = svd.fit_transform(x_train_tfidf)

# Mengubah svd_matrix menjadi DataFrame
df_svd = pd.DataFrame(svd_matrix, columns=[f"component_{i+1}" for i in range(n_components)])
df_svd
```

**Modeling**

---


```python
from sklearn.linear_model import LogisticRegression

# Latih model dengan beberapa parameter
logistic_model = LogisticRegression(max_iter=1000, C=0.2)
logistic_model.fit(svd_matrix, y_train)
```

**Evaluasi Model**

---



```python
from sklearn.metrics import accuracy_score, classification_report

# Prediksi menggunakan data uji
y_pred = logistic_model.predict(svd_matrix)

# Menghitung akurasi
accuracy = accuracy_score(y_train, y_pred)
print(f"Akurasi: {accuracy}")

# Menampilkan classification report untuk metrik precision, recall, dan F1-score
print("\nClassification Report:")
print(classification_report(y_train, y_pred))
```

```python
# Simpan TF-IDF vectorizer dan model
import pickle

with open('tfidf_vectorizer.pkl', 'wb') as tfidf_file:
    pickle.dump(vectorizer, tfidf_file)


with open('logistic_regression_model.pkl', 'wb') as model_file:
    pickle.dump(logistic_model, model_file)
```
