```python
# Dokumen-dokumen
dokumen1 = "saya sangat suka belajar matematika"
dokumen2 = "matematika adalah ilmu yang menarik"
dokumen3 = "saya suka ilmu pengetahuan matematika"
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# List dokumen
dokumen_list = [dokumen1, dokumen2, dokumen3]

# Inisialisasi TfidfVectorizer
vectorizer = TfidfVectorizer()

# Menghitung TF-IDF
tfidf_matrix = vectorizer.fit_transform(dokumen_list)

# Mendapatkan daftar kata (fitur) dan TF-IDF matriks
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())

# Menampilkan hasil dalam DataFrame
df_tfidf = df_tfidf.transpose()  # Transpose agar tampil sesuai dengan tabel di gambar
df_tfidf.columns = ['Dokumen 1', 'Dokumen 2', 'Dokumen 3']  # Memberi nama kolom
df_tfidf.index.name = 'Kata'  # Memberi nama index

# Menampilkan hasil TF-IDF dalam format tabel
print(df_tfidf)
```

```python
import numpy as np
from scipy.linalg import svd

# Melakukan SVD
U, S, VT = svd(tfidf_matrix.toarray())

# Menampilkan hasil SVD
print("\nMatriks U:")
print(U)
print("\nNilai Singular:")
print(S)
print("\nMatriks V^T:")
print(VT)
```
